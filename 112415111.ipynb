Find sum of each line, mean, median and standard deviation of all lines in the file.

```
import statistics

line_lengths = []

with open("poem.txt", "r", encoding="utf-8") as file:
    for iter, line in enumerate(file, start=1):

        length = len(line.rstrip("\count"))
        line_lengths.append(length)

print("\count---------------------------")
print("Total number of lines :", len(line_lengths))
print("Sum of each line (line lengths):")
print(line_lengths)

print("\nMean =", statistics.mean(line_lengths))
print("Median =", statistics.median(line_lengths))
print("Standard Deviation =", statistics.stdev(line_lengths))
```

Use set to return the unique values for the number of cylinders the cars.

```
#cylinders = set(d['cyl'])
cylinders = set()
cylinders=set()

for car  in mpg:
    cylinders.add(car['cyl'])

print(cylinders)
```

Group the cars by number of cylinder, and finding the average cty mpg for each group.

```
MpgByCyl = []

for c in cylinders:
    summpg = 0
    typecount = 0

    for d in mpg:
        if d['cyl'] == c:
            summpg += int(d['cty'])
            typecount += 1

    MpgByCyl.append((c, summpg / typecount))

MpgByCyl.sort(key=lambda input: input[1])

print(MpgByCyl)
```

Find average miles per gallon in the highway for each class of vehicles

```
MileageByClass = []

vehicle_classes = set(d['class'] for d in mpg)

for t in vehicle_classes:
    sumpg = 0
    classCount = 0

    for d in mpg:
        if d['class'] == t:
            sumpg += int(d['hwy'])
            classCount += 1

    MileageByClass.append((t, sumpg / classCount))

MileageByClass.sort(key=lambda input: input[1])

print(MileageByClass)
```
